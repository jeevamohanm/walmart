### Data Standardization

```python
df.columns= df.columns.str.lower()
```

```python
df
```

### Dropping duplicate values

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

### dropping null values

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

### Changing data type

```python
df['unit_price'] = df['unit_price'].str.replace('$' , '').astype(float)
df
```

```python
df.dtypes
```

### Feature engineering

```python
df['total'] = df['unit_price'] * df['quantity'] 
df
```

### Connecting to sql


```python
pip install pymysql sqlalchemy
```

```python
from sqlalchemy import create_engine,URL
```

```python
username = 'root'
password = input('enter your sql password')
url = URL.create("mysql+pymysql", username = username, password = password, host = 'localhost', port = 3306, database = 'walmart_db')
engine = create_engine(url)
with engine.connect() as conn:
    print('my sql is connected successfully')
```

```python
df.to_sql('walmart',engine, if_exists = 'replace', index=False)
print('dataset saved successfully')
```
